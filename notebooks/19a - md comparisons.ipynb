# Mimimum structures for Ag6

```python
isomer_names = ['planar', '3d']
ids = [2206361, 2206363, 2206362, 2206364]

ag6s = []

i = 0
for isomer_name in isomer_names:
    for mode in ['lcao', 'fd']:
        ag6s.append(read(get_project_path() + 'ml-dft/ag6_dft_calculations/structure_optimization/ag6_{:s}_{:s}_{:d}.traj'.format(isomer_name, mode, ids[i])))
        i += 1
```

```python
energies = [ag6.get_potential_energy() for ag6 in ag6s]
energies
```

```python
from flonacomldft.dft_calculator import DFTCalculator

calculator = DFTCalculator()
calculator.initialize_calculator()
```

```python
u = []

for i in range(2, len(ag6s)):
    molecule = ag6s[0].copy()
    u.append(calculator.calculate_potential_energy(
    molecule, 
    filename='ag6_{:d}.out'.format(i)
    ))
```
