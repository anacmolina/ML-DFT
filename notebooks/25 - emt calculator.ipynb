```python
from ase.calculators.emt import EMT
from ase.optimize import QuasiNewton
from ase.visualize.plot import plot_atoms
from flonacomldft.utils.silver_isomers_utils import get_molecule_isomer_minima
import matplotlib.pyplot as plt
import numpy as np
```

```python
def optimize_emt(init_config, label='INIT', add_diff=False):
    
    init_config.center(vacuum=5)
    init_config.calc = EMT()

    init_config.set_pbc(True)
    emt_config = init_config.copy()
    emt_config.calc = EMT()
    qn = QuasiNewton(emt_config)
    qn.run(0.001)
    
    if add_diff:
        nplots = 3
        large = 12
    else:
        nplots = 2
        large = 8
        
    fig, axs = plt.subplots(1, nplots, figsize=(large, 4))
    
    plot_atoms(init_config, ax=axs[0])
    axs[0].set_title(label)
    axs[0].text(0.95, 0.95, 'Epot: {:f}'.format(init_config.get_potential_energy()))
    
    plot_atoms(emt_config, ax=axs[1])
    axs[1].set_title('EMT')
    axs[1].text(0.95, 0.95, 'Epot: {:f}'.format(emt_config.get_potential_energy()))

    if add_diff:
        
        im = axs[2].imshow(init_config.get_all_distances()-emt_config.get_all_distances())
    
        fig.colorbar(im)
    
    return emt_config
```

```python
import torch
import numpy as np
from flonacomldft.utils.io_utils import get_path
from ase.md.langevin import Langevin
from ase.md.nvtberendsen import NVTBerendsen
from ase.md.verlet import VelocityVerlet
from ase.md.velocitydistribution import (MaxwellBoltzmannDistribution, 
                                         Stationary, ZeroRotation)
from ase.md.andersen import Andersen
from ase.io.trajectory import Trajectory
from ase import units
from flonacomldft.utils.plots import set_plot_iteration 


def run_velocity_verlet(atoms, dt, filename, iters, T=300):
    
    atoms = atoms.copy()
    atoms.calc = EMT()
    
    MaxwellBoltzmannDistribution(atoms, temperature_K=T)
    Stationary(atoms)  # zero linear momentum
    ZeroRotation(atoms)  # zero angular momentum

    dyn = VelocityVerlet(atoms, dt * units.fs)
    traj = Trajectory(filename, 'w', atoms)
    dyn.attach(traj.write, interval=1)
    
    dyn.run(iters)
    
def run_berendsen(atoms, dt, filename, iters, taut=50, T=300):
    
    atoms = atoms.copy()
    atoms.calc = EMT()

    Stationary(atoms)  # zero linear momentum
    ZeroRotation(atoms)  # zero angular momentum

    dyn = NVTBerendsen(atoms, dt * units.fs, T, taut=taut*units.fs)
    traj = Trajectory(filename, 'w', atoms)
    dyn.attach(traj.write, interval=1)
    
    dyn.run(iters)
    
def run_langevin(atoms, dt, filename, iters, friction=0.002, T=300):
    
    atoms = atoms.copy()
    atoms.calc = EMT()

    Stationary(atoms)  # zero linear momentum
    ZeroRotation(atoms)  # zero angular momentum

    dyn = Langevin(atoms, dt * units.fs, temperature_K=T, friction=friction)
    traj = Trajectory(filename, 'w', atoms)
    dyn.attach(traj.write, interval=1)
    
    dyn.run(iters)

def run_andersen(atoms, dt, filename, iters, prob=0.002, T=300):
    
    atoms = atoms.copy()
    atoms.calc = EMT()

    Stationary(atoms)  # zero linear momentum
    ZeroRotation(atoms)  # zero angular momentum

    dyn = Andersen(atoms, dt * units.fs, temperature_K=T, andersen_prob=prob)
    traj = Trajectory(filename, 'w', atoms)
    dyn.attach(traj.write, interval=1)
    
    dyn.run(iters)
```

```python
def analyze_md(traj):
    
    fig, axs = plt.subplots(1, 5, figsize=(20, 4))
    axs = axs.ravel()
    for i, ax in zip(np.linspace(0, len(traj)-1, 5), axs):
        plot_atoms(traj[int(i)], ax)
        
    temperature =  torch.tensor([config.get_temperature() for config in traj])

    fig, axs = plt.subplots(1, 2, figsize=(16, 4))
    set_plot_iteration(temperature, avg=True, window_size=300, label='temperature', ax=axs[0])
    axs[0].set_xlabel('steps')
    axs[0].set_ylabel('Temperature')
    
    epot =  torch.tensor([config.get_potential_energy() for config in traj])

    set_plot_iteration(epot, avg=True, window_size=250, label='temperature', ax=axs[1])
    axs[1].set_xlabel('steps')
    axs[1].set_ylabel('Epot') 
```
