```python
import os

import torch
import numpy as np

from flonacomldft.gaussian_utils import MoG
import matplotlib.pyplot as plt

from flonacomldft.models.real_nvp import RealNVP_MLP
from flonacomldft.train_flow_from_data import train_flow

from flonacomldft.full_adaptive_sampling import adaptive_sampling

from flonacomldft.utils.diagnostics import avg_windows
from flonacomldft.utils.plots import set_plot_sequential_data

from ase.units import kB

import warnings
warnings.filterwarnings('ignore')

import pandas as pd
import json
```

```python
device = 'cpu'

sep = 1.
means = [torch.tensor([-sep, 0]), torch.tensor([sep, 0])]
covars = [2e-1 * torch.eye(2),  2e-1 * torch.eye(2)]

target = MoG(means, covars, weights=[0.2, 0.8])
```

```python
n_samples = 10000

target_samples = target.sample(n_samples)
target_us = target.U(target_samples)
id_mode = torch.zeros(n_samples)

dataset = torch.cat([target_samples.T, target_us.reshape(1, -1), id_mode.reshape(1, -1)]).T

train = dataset[:int(n_samples*0.8)]
test = dataset[int(n_samples*0.8):]
```

```python
T = 300
target_energy = lambda x: target.U(x) / (1/(kB*T))
mlp_dic = {'model': target_energy}
```

```python
n_blocks = 4
hidden_dim = 64
hidden_depth = 4

model = RealNVP_MLP(2,
                    n_blocks=n_blocks,
                    block_depth=1,
                    init_weight_scale=1e-3,
                    base_cov=None,
                    hidden_dim=hidden_dim,
                    hidden_depth=hidden_depth,
                    device=device,
                    )

flow_dic = train_flow(
    model,
    train,
    test,
    dim=2,
    n_iter=50,
    lr=1e-3,
    use_scheduler=False,
    step_schedule=100,
    save_splits=10,
    grad_clip=1e4,
    with_tqdm=False,
    compute_part_ratio=True,
    energy_type='mlp',
    mlp_model=target.U,
    n_prop=100,
    batch_size=8000,
)
```

```python
flow_hyperparams = {'n_iter': 100,
    'lr': 1e-3,
    'use_scheduler': False,
    'step_schedule': 100,
    'save_splits': 10,
    'grad_clip': 1e4,
    'compute_part_ratio': True,
    'n_prop': 100,
    'energy_type': 'mlp',
    'batch_size': 8000,
    }

# MCMC Params
n_runs = 10
n_chains = 50
n_steps = 50

# Init chains
init_samples = target.sample(n_chains)
init_us = target.U(init_samples)
init_id_mode = torch.zeros(n_chains)

init_mcmc = torch.cat([init_samples.T, init_us.reshape(1, -1), init_id_mode.reshape(1, -1)]).T
```

```python
_ = adaptive_sampling([train], 
                         [test],
                         init_mcmc, 
                         dim=2,
                         dict_flows_init=[flow_dic],
                         dict_mlps_init=[mlp_dic],
                         mixture=False,
                         flow_hyperparams=[flow_hyperparams],
                         n_runs=n_runs,
                         n_chains=n_chains,
                         n_steps=n_steps,
                         path=os.getcwd()
                        )
```

```python
lims = {'x_min':-2 * sep, 'x_max':2 * sep, 'y_min':-2 * sep, 'y_max':2 * sep}
device = 'cpu'

## TODO: add to gaussian utils

def set_plot(ax, n_points=1000, target_log_prob=target.log_prob, prop_log_prob=None, lims=lims):
    
    x_range = torch.linspace(lims['x_min'], lims['x_max'], n_points, device=device)

    if lims['y_min'] is None:
        y_range = x_range.clone()
    else:
        y_range = torch.linspace(lims['y_min'], lims['y_max'], n_points, device=device)

    grid = torch.meshgrid(x_range, y_range)
    xys = torch.stack(grid).reshape(2, n_points ** 2).T.to(device)

    Us_target = target_log_prob(xys).reshape(n_points, n_points).T.detach().cpu().numpy()

    if ax is None:
        plt.figure()
    else:
        plt.sca(ax)
    #plt.axis('off')

    plt.contourf(x_range, y_range, np.exp(Us_target)
    # np.log( - Us_target + Us_target.max() + 1)
    , 10, cmap='GnBu')
    ax.set_aspect('equal')

    if prop_log_prob is not None:
        Us_prop = prop_log_prob(xys).reshape(n_points, n_points).T.detach().cpu().numpy()

        plt.contour(x_range, y_range, np.exp(Us_prop), 10, colors='k', linestyles=':', alpha=0.5)
```

```python
# EMT Results

def load_simulation_info(id, i, isomer, add_part_ratios=False, path=os.getcwd()):
    id_name = "{:d}{:d}".format(id, i)
    
    file_accs = path + "results_adaptive_is{:d}_{:s}/accs_is{:d}_{:s}.csv".format(isomer, id_name, isomer, id_name)
    df_acc = pd.read_csv(file_accs)
    accs = df_acc[["accs", "time"]].to_numpy()
    
    if add_part_ratios:
        file_part_ratios = path + "results_adaptive_is{:d}_{:s}/part_ratios_is{:d}_{:s}.csv".format(isomer, id_name, isomer, id_name)
        df_part_ratios = pd.read_csv(file_part_ratios)
        part_ratios = df_part_ratios[["part_ratios", "time"]].to_numpy()

    file_args = path + "results_adaptive_is{:d}_{:s}/args_adaptive_sampling_{:s}.json".format(isomer,id_name, id_name)
    f = open(file_args)
    args = json.load(f)
    f.close()
    
    if add_part_ratios:
        return accs, args, part_ratios
    else:
        return accs, args
    

def get_information(ids, path, n_sim, isomer, add_part_ratios=False):
    
    if add_part_ratios:
        all_part_ratios = []
    
    all_accs = []
    all_args = []

    for id_ in ids:
        for j in range(0, n_sim):
            data = load_simulation_info(id_, j, isomer=isomer, path=path, add_part_ratios=add_part_ratios)
            all_accs.append(data[0])
            all_args.append(data[1])
            
            if add_part_ratios:
                all_part_ratios.append(data[2])

    md = np.unique([args["folder_path"] for args in all_args])

    hyperparams = ['learning_rate', 'n_iter', 'batch_size', 'n_blocks', 'hidden_dim', 'hidden_depth', 'n_runs', 'n_chains', 'n_steps']
    hyperparams_values = [[] for i in range(len(hyperparams))]

    for i in range(len(hyperparams)):
        for args in all_args:
            hyperparams_values[i].append(args[hyperparams[i]])
            
    hyperparams_values = [np.unique(hyperparams_values[i]) for i in range(len(hyperparams))]
    
    to_return = {'all_accs': all_accs,
                 'all_args': all_args,
                 'hyperparams': hyperparams,
                 'hyperparams_values': hyperparams_values,
                 'isomer': isomer,
                 'md': md,
                }
    
    if add_part_ratios:
        to_return['all_part_ratios'] = all_part_ratios

    
    return to_return

def plot_hyperparams_search_with_ratios(simulation_data, key, ylabel, window=100, upper_bound=0.0):
    
    isomer = simulation_data['isomer']
    md = simulation_data['md']
    all_values = simulation_data[key]
    all_args = simulation_data['all_args']
    hyperparams = simulation_data['hyperparams']
    hyperparams_values = simulation_data['hyperparams_values']
    
    fig, axs = plt.subplots(3, 3, figsize=(20, 16))
    fig.subplots_adjust(top=0.92)

    axs = axs.ravel()

    for i in range(len(all_values)):
    
        x = all_values[i][:, 1]/3600
        y = all_values[i][:, 0]

        x_avg = avg_windows(x, window, axis=1)
        y_avg = avg_windows(y, window, axis=1)

        if y_avg.max() > upper_bound:

            for j in range(len(hyperparams)):
                c = cm.jet(np.linspace(0.1, 0.9, len(hyperparams_values[j])))
                mask = hyperparams_values[j] == all_args[i][hyperparams[j]]
                axs[j].plot(x_avg, y_avg, c=c[mask][0], alpha=0.6, label=hyperparams[j]+"="+str(all_args[i][hyperparams[j]]))

                axs[j].set_title(hyperparams[j])
                axs[j].set_xlabel("Time (h)")
                axs[j].set_ylabel(ylabel)

                handles, labels = axs[j].get_legend_handles_labels()
                by_label = dict(zip(labels, handles))
                axs[j].legend(by_label.values(), by_label.keys())

    fig.suptitle("Adaptive MCMC params: \n isomer={} md={}".format(isomer,
                                                                md[0],) )
    
    return fig, axs
```
