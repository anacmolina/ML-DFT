```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

import warnings
warnings.filterwarnings("ignore")
```

```python
log_weigths = np.load('/home/ana/Downloads/logweights.npy')
samples_from_flow = np.load('/home/ana/Downloads/samples_from_flow.npy')
```

```python
from flonacomldft.utils.io_utils import get_path
from flonacomldft.FES.plotter2 import Plotter

plotter = Plotter(400, 'Ag6')
plotter.readfile(get_path() + 'unrotated_300.txt')

fig, ax = plotter.plot_fes(0.1, 300, delta2=1, shift=1.5)
ax.scatter(samples_from_flow[0][:, 0], samples_from_flow[0][:, 1], c=log_weigths[0], cmap='autumn')
ax.scatter(samples_from_flow[1][:, 0], samples_from_flow[1][:, 1], c=log_weigths[1], cmap='autumn')
fig.set_size_inches(10, 6)
size = 12
ax.xaxis.label.set_size(size)
ax.yaxis.label.set_size(size)
plt.xticks(fontsize=size)
plt.yticks(fontsize=size);
```

```python
all_samples = samples_from_flow.reshape(400, 2)
all_log_weights = log_weigths.reshape(400)
import pandas as pd
df = pd.DataFrame(all_samples, columns=['C', 'R'])
```

```python
from flonacomldft.utils.io_utils import get_path
from flonacomldft.FES.plotter2 import Plotter

plotter = Plotter(400, 'Ag6')
plotter.readfile(get_path() + 'unrotated_300.txt')

fig, ax = plotter.plot_fes(0.1, 300, delta2=1, shift=1.5)
sns.kdeplot(samples_from_flow[0][:, 0], samples_from_flow[0][:, 1], weights=log_weigths[0], cmap='autumn', ax=ax)
sns.kdeplot(samples_from_flow[1][:, 0], samples_from_flow[1][:, 1], weights=log_weigths[1], cmap='autumn', ax=ax)
fig.set_size_inches(10, 6)
size = 12
ax.xaxis.label.set_size(size)
ax.yaxis.label.set_size(size)
plt.xticks(fontsize=size)
plt.yticks(fontsize=size);
```
