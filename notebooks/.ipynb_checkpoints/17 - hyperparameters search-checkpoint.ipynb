```python
from flonacomldft.utils.io_utils import load_pickle_file, get_project_path
import numpy as np
import matplotlib.pyplot as plt
```

```python
def load_losses(model, isomer, ID, i):
    if isomer==0:
        name='planar'
    elif isomer==1:
        name='3d'

    losses = load_pickle_file('training/'+name+'/'+str(model)+'/results/is'+str(isomer)+'_'+str(model)[:-1]+'_dic_training_'+str(ID)+'_'+str(i)+'.pkl', 
                            path=get_project_path())['losses']
    
    return np.array(losses).T[-1]
```

```python
id_flows_0 = 2147988 + np.arange(0,5)
id_mlps_0 = 2148010 + np.arange(0,5)

id_flows_1 = 2147994 + np.arange(0,5)
id_mlps_1 = 2148015 + np.arange(0,5)
```

```python
losses_flow_0 = [np.array([load_losses('flows', 0, id_flows_0[j], i) for i in range(64)]) for j in range(5)]
```

```python
for j in range(5):
    plt.figure(figsize=(15, 5))
    plt.plot(losses_flow_0[j], label=['train', 'test'])
    plt.legend()
    plt.ylabel('losses')
    plt.xlabel('hidden dim nodes')
    plt.title('PLANAR FLOW hidden depth {:d}'.format(4*(j+1)))
    plt.xticks(ticks=np.arange(0, 64, 8), labels=np.arange(4, 260, 32));
```

```python
losses_flow_1 = [np.array([load_losses('flows', 1, id_flows_1[j], i) for i in range(64)]) for j in range(5)]
```

```python
for j in range(5):
    plt.figure(figsize=(15, 5))
    plt.plot(losses_flow_1[j], label=['train', 'test'])
    plt.legend()
    plt.ylabel('losses')
    plt.xlabel('hidden dim nodes')
    plt.title('3D FLOW hidden depth {:d}'.format(4*(j+1)))
    plt.xticks(ticks=np.arange(0, 64, 8), labels=np.arange(4, 260, 32));
```

```python
losses_mlps_0 = [np.array([load_losses('mlps', 0, id_mlps_0[j], i) for i in range(64)]) for j in range(5)]
```

```python
for j in range(5):
    plt.figure(figsize=(15, 5))
    plt.plot(losses_mlps_0[j], label=['train', 'test'])
    plt.yscale('log')
    plt.legend()
    plt.ylabel('losses')
    plt.xlabel('hidden dim nodes')
    plt.title('PLANAR MLP hidden depth {:d}'.format(4*(j+1)))
    plt.xticks(ticks=np.arange(0, 64, 8), labels=np.arange(4, 260, 32));
```
