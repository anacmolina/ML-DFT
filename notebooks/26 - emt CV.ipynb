```python
from flonacomldft.collective_variables import compute_C, compute_R
from flonacomldft.utils.io_utils import get_path
import matplotlib.pyplot as plt
import numpy as np
from ase.io.trajectory import Trajectory
```

```python
isomers_traj_bd = [Trajectory(get_path()+'/is'+str(i)+'_emt_berendsen.traj') for i in range(4)]
```

```python
cvs = []
for j in range(len(isomers_traj_bd)):
    cv = []
    for i in range(len(isomers_traj_bd[j])):
        cv.append([compute_C(isomers_traj_bd[j][i]), compute_R(isomers_traj_bd[j][i])])
    cvs.append(cv)
```

```python
cvs = np.array(cvs)
```

```python
from flonacomldft.utils.io_utils import get_path
from flonacomldft.FES.plotter2 import Plotter

n = 4500

plotter = Plotter(400, 'Ag6')
plotter.readfile(get_path()+'/unrotated_300.txt')

fig, ax = plotter.plot_fes(0.1, 300, delta2=1, shift=1.5)
ax.scatter(cvs[0][n:, 0], cvs[0][n:, 1], label='is0')
ax.scatter(cvs[1][n:, 0], cvs[1][n:, 1], label='is1')
ax.scatter(cvs[2][n:, 0], cvs[2][n:, 1], label='is2')
ax.scatter(cvs[3][n:, 0], cvs[3][n:, 1], label='is3')
#fig.set_size_inches(10, 6)
size = 12
ax.xaxis.label.set_size(size)
ax.yaxis.label.set_size(size)
plt.xticks(fontsize=size)
plt.yticks(fontsize=size)
plt.legend()
plt.title('From {:d} step'.format(n));
```
